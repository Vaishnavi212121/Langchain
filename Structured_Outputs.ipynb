```python
from langchain_groq import ChatGroq
#from langchain_core.messages import HumanMessage,SystemMessage,AIMessage
from langchain_core.prompts import PromptTemplate
import os
from google.colab import userdata
os.environ["GROQ_API_KEY"] =  userdata.get('GROQ_API_KEY')
```

```python
llm_groq = ChatGroq(model="qwen/qwen3.8-27b",temperature=0,max_tokens=200)
llm_groq.invoke("Tell me a joke.Generate the o/p in key value pair format ")
```

#Structured Output Using Pydantic Models

```python
from pydantic import BaseModel,Field

class llm_schema(BaseModel):
  setup:str = Field(description="The setup for Joke")
  punchline:str = Field(description="The punchline")
```

```python
llm_structured_output = llm_groq.with_structured_output(llm_schema)

llm_structured_output.invoke("tell me a Joke")
```
