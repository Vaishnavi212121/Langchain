```python
import os
from google.colab import userdata
os.environ["GROQ_API_KEY"]=userdata.get("GROQ_API_KEY")
from langchain_groq import ChatGroq

# Initialize ChatGroq with a max_tokens limit to prevent large outputs
client = ChatGroq(api_key=os.environ.get("GROQ_API_KEY"),model ="qwen/qwen3.8-27b", max_tokens=100)
chat_completion = client.invoke(
    input = [
      {"role":"user","content":"Tell me a fun fact"}
    ]
)
print(chat_completion.content)
```
