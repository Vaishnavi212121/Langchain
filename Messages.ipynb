##Messages

```python
from langchain_core.messages import HumanMessage,SystemMessage,AIMessage

from langchain_groq import ChatGroq
import os
from google.colab import userdata
os.environ["GROQ_API_KEY"]=userdata.get('GROQ_API_KEY')

llm_groq = ChatGroq(model="qwen/qwen3.8-27b",temperature=0,max_tokens=200)

my_messages = [
    SystemMessage(content="Your are a helpful assistant"),
    HumanMessage(content="What is the capital of India?")
]
llm_groq.invoke(my_messages).content
# AIMessage(content="The capital of India is New Delhi")
```

##**Prompts**

```python
from langchain_core.prompts import PromptTemplate

user_input = input("Enter the country to know the capital:")

dynamic_prompt = PromptTemplate.from_template("What is the capital of {country}?") #country will come from user

new_prompt = dynamic_prompt.invoke({"country":user_input})
llm_groq.invoke(new_prompt).content
```

```python
from langchain_core.prompts import ChatPromptTemplate
prompt_template = ChatPromptTemplate.from_messages([
    ("system", "Your are a helpful assistant"),
    ("user", "What is capital of {country}")
])
user_input = input("Enter the country to know the capital:")
ready_prompt = prompt_template.invoke({"country":user_input})

#ready_prompt

llm_groq.invoke(ready_prompt).content
```
